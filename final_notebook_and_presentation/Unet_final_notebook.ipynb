```python
def evaluate_and_visualize_with_weights(model, test_loader, device, num_samples=5):
    """
    Evaluate the model on the test dataset and visualize predictions along with weight maps.
    """
    model.eval()
    with torch.no_grad():
        for i, (inputs, labels, weights) in enumerate(test_loader):
            if i >= num_samples:
                break

            inputs, labels, weights = inputs.to(device), labels.to(device), weights.to(device)
            outputs = model(inputs)
            predictions = torch.sigmoid(outputs) > 0.5

            # Преобразование в numpy
            inputs_np = inputs.cpu().squeeze().numpy()
            labels_np = labels.cpu().squeeze().numpy()
            weights_np = weights.cpu().squeeze().numpy()
            predictions_np = predictions.cpu().squeeze().numpy()

            # Построение графиков
            plt.figure(figsize=(20, 5))

            # Исходное изображение
            plt.subplot(1, 4, 1)
            plt.imshow(inputs_np, cmap="gray")
            plt.title("Input Image")
            plt.axis("off")

            # Метки
            plt.subplot(1, 4, 2)
            plt.imshow(labels_np, cmap="gray")
            plt.title("Ground Truth")
            plt.axis("off")

            # Карта весов
            plt.subplot(1, 4, 3)
            plt.imshow(weights_np, cmap="viridis")
            plt.title("Weight Map")
            plt.axis("off")

            # Предсказания
            plt.subplot(1, 4, 4)
            plt.imshow(predictions_np, cmap="gray")
            plt.title("Prediction")
            plt.axis("off")

            plt.show()
```

```python
import os
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from tqdm.notebook import tqdm
from PIL import Image
import cv2
import tifffile as tiff
import matplotlib.pyplot as plt


# --------------------- Загрузка данных ---------------------
def read_tiff_data(volume_path, labels_path):
    volume = tiff.imread(volume_path)  # Shape: (N, 512, 512)
    labels = tiff.imread(labels_path)  # Shape: (N, 512, 512)
    labels = (labels > 0).astype(np.uint8)  # Binarize labels
    return volume, labels


# --------------------- Weight Map Generation ---------------------
def weight_map(mask, w0=10, sigma=5):
    """
    Генерация карты весов для компенсации дисбаланса классов и границ объектов.
    """
    mask = mask.astype(np.uint8)
    _, regions = cv2.connectedComponents(mask)
    distances = np.zeros((mask.shape[0], mask.shape[1], regions.max() + 1))
    for i in range(1, regions.max() + 1):
        region_mask = (regions == i).astype(np.uint8)
        distances[..., i] = cv2.distanceTransform(1 - region_mask, cv2.DIST_L2, 0)
    combined_distances = np.min(distances[..., 1:], axis=-1)
    weight = w0 * np.exp(-(combined_distances**2) / (2 * sigma**2))
    weight[mask == 1] += 1
    return weight


# --------------------- U-Net Architecture ---------------------
class Unet(nn.Module):
    def __init__(self, channels, no_classes, output_size=None):
        super(Unet, self).__init__()
        self.output_size = output_size
        self.encoder = Encoder(channels)
        dec_channels = list(reversed(channels[1:]))
        self.decoder = Decoder(dec_channels)
        self.head = nn.Conv2d(channels[1], no_classes, kernel_size=1)

    def forward(self, x):
        x, features = self.encoder(x)
        x = self.decoder(x, features)
        x = self.head(x)
        if self.output_size:
            x = nn.functional.interpolate(x, size=self.output_size, mode="bilinear", align_corners=False)
        return x


class Encoder(nn.Module):
    def __init__(self, channels):
        super(Encoder, self).__init__()
        self.blocks = nn.ModuleList([
            Block(in_channel, out_channel)
            for in_channel, out_channel in zip(channels[:-1], channels[1:])
        ])
        self.pool = nn.MaxPool2d(2)

    def forward(self, x):
        features = []
        for block in self.blocks[:-1]:
            x = block(x)
            features.append(x)
            x = self.pool(x)
        x = self.blocks[-1](x)
        return x, features


class Decoder(nn.Module):
    def __init__(self, channels):
        super(Decoder, self).__init__()
        self.upconvs = nn.ModuleList([
            nn.ConvTranspose2d(in_channel, out_channel, kernel_size=2, stride=2)
            for in_channel, out_channel in zip(channels[:-1], channels[1:])
        ])
        self.blocks = nn.ModuleList([
            Block(in_channel, out_channel)
            for in_channel, out_channel in zip(channels[:-1], channels[1:])
        ])

    def forward(self, x, features):
        for upconv, block, feature in zip(self.upconvs, self.blocks, reversed(features)):
            x = upconv(x)
            feature = self.center_crop(feature, x.shape[2], x.shape[3])
            x = torch.cat([x, feature], dim=1)
            x = block(x)
        return x

    @staticmethod
    def center_crop(tensor, target_height, target_width):
        _, _, height, width = tensor.size()
        delta_h = (height - target_height) // 2
        delta_w = (width - target_width) // 2
        return tensor[:, :, delta_h:delta_h + target_height, delta_w:delta_w + target_width]


class Block(nn.Module):
    def __init__(self, in_channel, out_channel, kernel_size=3):
        super(Block, self).__init__()
        self.conv1 = nn.Conv2d(in_channel, out_channel, kernel_size, padding=1)
        self.conv2 = nn.Conv2d(out_channel, out_channel, kernel_size, padding=1)
        self.relu = nn.ReLU(inplace=True)

    def forward(self, x):
        x = self.relu(self.conv1(x))
        x = self.relu(self.conv2(x))
        return x


# --------------------- Dataset and DataLoader ---------------------
class SegmentationDataset(torch.utils.data.Dataset):
    def __init__(self, images, masks):
        self.images = images
        self.masks = masks

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        image = self.images[idx]
        mask = self.masks[idx]
        # Normalize images
        image = (image - image.min()) / (image.max() - image.min())
        weight = weight_map(mask)
        image = np.expand_dims(image, axis=0)
        mask = np.expand_dims(mask, axis=0)
        weight = np.expand_dims(weight, axis=0)
        return (
            torch.tensor(image, dtype=torch.float32),
            torch.tensor(mask, dtype=torch.float32),
            torch.tensor(weight, dtype=torch.float32),
        )


# --------------------- Weighted Loss ---------------------
class WeightedBCEWithLogitsLoss(nn.Module):
    def __init__(self):
        super(WeightedBCEWithLogitsLoss, self).__init__()

    def forward(self, inputs, targets, weights):
        bce = nn.functional.binary_cross_entropy_with_logits(inputs, targets, reduction="none")
        weighted_bce = weights * bce
        return weighted_bce.mean()


# --------------------- Training Setup ---------------------
def train_unet(model, train_loader, criterion, optimizer, device, epochs=50):
    model.to(device)
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for inputs, labels, weights in tqdm(train_loader):
            inputs, labels, weights = inputs.to(device), labels.to(device), weights.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels, weights)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()
        print(f"Epoch {epoch + 1}/{epochs}, Loss: {train_loss / len(train_loader)}")


# --------------------- Main Code ---------------------
if __name__ == "__main__":
    # File paths
    train_vol_path = "./data/train-volume.tif"
    train_label_path = "./data/train-labels.tif"
    test_vol_path = "./data/test-volume.tif"
    test_label_path = "./data/test-labels.tif"

    # Load data
    train_volume, train_labels = read_tiff_data(train_vol_path, train_label_path)
    test_volume, test_labels = read_tiff_data(test_vol_path, test_label_path)

    # Prepare datasets
    train_dataset = SegmentationDataset(train_volume, train_labels)
    test_dataset = SegmentationDataset(test_volume, test_labels)

    train_loader = DataLoader(train_dataset, batch_size=1, shuffle=True)
    test_loader = DataLoader(test_dataset, batch_size=1)

    # Create model
    model = Unet([1, 64, 128, 256, 512, 1024], no_classes=1, output_size=(512, 512))
    optimizer = optim.SGD(model.parameters(), lr=1e-3, momentum=0.99)
    criterion = WeightedBCEWithLogitsLoss()

    # Train model
    train_unet(model, train_loader, criterion, optimizer, device="cuda", epochs=500)
    evaluate_and_visualize_with_weights(model, test_loader, device="cuda", num_samples=5)
```

```python
def visualize_augmentation(dataset, num_samples=3, w0=10, sigma=5):
    """
    Визуализация оригинальных и аугментированных данных.
    """
    for i in range(num_samples):
        idx = np.random.randint(0, len(dataset))
        sample = dataset[idx]

        # Исходные данные
        original_image = dataset.images[idx]
        original_mask = dataset.masks[idx]

        # Аугментированные данные
        augmented_image = sample[0].numpy().squeeze()
        augmented_mask = sample[1].numpy().squeeze()
        weights = sample[2].numpy().squeeze()

        # Построение графиков
        fig, ax = plt.subplots(3, 3, figsize=(12, 12))

        ax[0, 0].imshow(original_image, cmap="gray")
        ax[0, 0].set_title("Original Image")
        ax[0, 0].axis("off")

        ax[0, 1].imshow(original_mask, cmap="gray")
        ax[0, 1].set_title("Original Mask")
        ax[0, 1].axis("off")

        ax[0, 2].imshow(original_image, cmap="gray")
        ax[0, 2].imshow(original_mask, cmap="jet", alpha=0.3)
        ax[0, 2].set_title("Overlay: Original")
        ax[0, 2].axis("off")

        ax[1, 0].imshow(augmented_image, cmap="gray")
        ax[1, 0].set_title("Augmented Image")
        ax[1, 0].axis("off")

        ax[1, 1].imshow(augmented_mask, cmap="gray")
        ax[1, 1].set_title("Augmented Mask")
        ax[1, 1].axis("off")

        ax[1, 2].imshow(augmented_image, cmap="gray")
        ax[1, 2].imshow(augmented_mask, cmap="jet", alpha=0.3)
        ax[1, 2].set_title("Overlay: Augmented")
        ax[1, 2].axis("off")

        ax[2, 0].hist(augmented_image.flatten(), bins=50, color="gray")
        ax[2, 0].set_title("Augmented Image Histogram")

        ax[2, 1].imshow(weights, cmap="viridis")
        ax[2, 1].set_title("Weight Map")
        ax[2, 1].axis("off")

        ax[2, 2].hist(augmented_mask.flatten(), bins=50, color="blue")
        ax[2, 2].set_title("Augmented Mask Histogram")

        plt.tight_layout()
        plt.show()
```

```python
import os
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from tqdm.notebook import tqdm
from PIL import Image
import cv2
import tifffile as tiff
import matplotlib.pyplot as plt
import albumentations as A
from albumentations.pytorch import ToTensorV2

# --------------------- Загрузка данных ---------------------
def read_tiff_data(volume_path, labels_path):
    volume = tiff.imread(volume_path)  # Shape: (N, 512, 512)
    labels = tiff.imread(labels_path)  # Shape: (N, 512, 512)
    labels = (labels > 0).astype(np.uint8)  # Binarize labels
    return volume, labels

# --------------------- Аугментация данных ---------------------
def get_train_transform():
    p = 0.95
    return A.Compose([
        A.OneOf([
            A.HorizontalFlip(p=p),
            A.VerticalFlip(p=p),
            A.Transpose(p=p),
            A.RandomRotate90(p=p),
            A.ShiftScaleRotate(p=p, shift_limit=0.0625, scale_limit=0.1, rotate_limit=45)
        ], p=1),
        A.GaussNoise(p=p, var_limit=(0, 20), mean=0, per_channel=True),
        A.MultiplicativeNoise(p=p, multiplier=(0.9, 1.1), elementwise=True),
        A.ElasticTransform(p=p, alpha=35, sigma=5, approximate=True),
        A.RandomBrightnessContrast(p=p, brightness_limit=0.15, contrast_limit=0.15),
        A.Normalize(mean=0.0, std=1.0),
        ToTensorV2()
    ])

# --------------------- Weight Map Generation ---------------------
def weight_map(mask, w0=10, sigma=5):
    mask = mask.astype(np.uint8)
    _, regions = cv2.connectedComponents(mask)
    distances = np.zeros((mask.shape[0], mask.shape[1], regions.max() + 1))
    for i in range(1, regions.max() + 1):
        region_mask = (regions == i).astype(np.uint8)
        distances[..., i] = cv2.distanceTransform(1 - region_mask, cv2.DIST_L2, 0)
    combined_distances = np.min(distances[..., 1:], axis=-1)
    weight = w0 * np.exp(-(combined_distances**2) / (2 * sigma**2))
    weight[mask == 1] += 1
    return weight

# --------------------- U-Net Architecture ---------------------
class Unet(nn.Module):
    def __init__(self, channels, no_classes, output_size=None):
        super(Unet, self).__init__()
        self.output_size = output_size
        self.encoder = Encoder(channels)
        dec_channels = list(reversed(channels[1:]))
        self.decoder = Decoder(dec_channels)
        self.head = nn.Conv2d(channels[1], no_classes, kernel_size=1)

    def forward(self, x):
        x, features = self.encoder(x)
        x = self.decoder(x, features)
        x = self.head(x)
        if self.output_size:
            x = nn.functional.interpolate(x, size=self.output_size, mode="bilinear", align_corners=False)
        return x

class Encoder(nn.Module):
    def __init__(self, channels):
        super(Encoder, self).__init__()
        self.blocks = nn.ModuleList([
            Block(in_channel, out_channel)
            for in_channel, out_channel in zip(channels[:-1], channels[1:])
        ])
        self.pool = nn.MaxPool2d(2)

    def forward(self, x):
        features = []
        for block in self.blocks[:-1]:
            x = block(x)
            features.append(x)
            x = self.pool(x)
        x = self.blocks[-1](x)
        return x, features

class Decoder(nn.Module):
    def __init__(self, channels):
        super(Decoder, self).__init__()
        self.upconvs = nn.ModuleList([
            nn.ConvTranspose2d(in_channel, out_channel, kernel_size=2, stride=2)
            for in_channel, out_channel in zip(channels[:-1], channels[1:])
        ])
        self.blocks = nn.ModuleList([
            Block(in_channel, out_channel)
            for in_channel, out_channel in zip(channels[:-1], channels[1:])
        ])

    def forward(self, x, features):
        for upconv, block, feature in zip(self.upconvs, self.blocks, reversed(features)):
            x = upconv(x)
            feature = self.center_crop(feature, x.shape[2], x.shape[3])
            x = torch.cat([x, feature], dim=1)
            x = block(x)
        return x

    @staticmethod
    def center_crop(tensor, target_height, target_width):
        _, _, height, width = tensor.size()
        delta_h = (height - target_height) // 2
        delta_w = (width - target_width) // 2
        return tensor[:, :, delta_h:delta_h + target_height, delta_w:delta_w + target_width]

class Block(nn.Module):
    def __init__(self, in_channel, out_channel, kernel_size=3):
        super(Block, self).__init__()
        self.conv1 = nn.Conv2d(in_channel, out_channel, kernel_size, padding=1)
        self.conv2 = nn.Conv2d(out_channel, out_channel, kernel_size, padding=1)
        self.relu = nn.ReLU(inplace=True)

    def forward(self, x):
        x = self.relu(self.conv1(x))
        x = self.relu(self.conv2(x))
        return x

# --------------------- Dataset and DataLoader ---------------------
class SegmentationDataset(torch.utils.data.Dataset):
    def __init__(self, images, masks, transform=None):
        self.images = images
        self.masks = masks
        self.transform = transform

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        image = self.images[idx]
        mask = self.masks[idx]

        if self.transform:
            augmented = self.transform(image=image, mask=mask)
            image = augmented["image"]
            mask = augmented["mask"]

        weight = weight_map(mask.numpy())
        return image.float(), torch.tensor(mask, dtype=torch.float32), torch.tensor(weight, dtype=torch.float32)

# --------------------- Weighted Loss ---------------------
class WeightedBCEWithLogitsLoss(nn.Module):
    def __init__(self):
        super(WeightedBCEWithLogitsLoss, self).__init__()

    def forward(self, inputs, targets, weights):
        bce = nn.functional.binary_cross_entropy_with_logits(inputs, targets, reduction="none")
        weighted_bce = weights * bce
        return weighted_bce.mean()

# --------------------- Training ---------------------
def train_unet(model, train_loader, criterion, optimizer, device, epochs, save_path, start_epoch=0):
    """
    Train the U-Net model, with an option to continue training from a specific epoch.
    """
    model.to(device)
    for epoch in range(start_epoch, epochs):
        model.train()
        train_loss = 0
        for inputs, labels, weights in tqdm(train_loader):
            inputs, labels, weights = inputs.to(device), labels.to(device), weights.to(device)

            # Add channel dimension to labels
            labels = labels.unsqueeze(1)

            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels, weights)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()
        print(f"Epoch {epoch + 1}/{epochs}, Loss: {train_loss / len(train_loader)}")

        # Save the model's state and optimizer's state for resuming training
        torch.save({
            'epoch': epoch + 1,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'loss': train_loss / len(train_loader),
        }, save_path)


# --------------------- Main ---------------------
if __name__ == "__main__":
    train_vol_path = "./data/train-volume.tif"
    train_label_path = "./data/train-labels.tif"
    weights_path = "./model_weights.pth"
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    train_volume, train_labels = read_tiff_data(train_vol_path, train_label_path)

    train_transform = get_train_transform()
    train_dataset = SegmentationDataset(train_volume, train_labels, transform=train_transform)
    train_loader = DataLoader(train_dataset, batch_size=1, shuffle=True)

    # Augmentatiom vizualization
    print("Visualizing augmentations...")
    visualize_augmentation(train_dataset, num_samples=3)

    # Model
    model = Unet([1, 64, 128, 256, 512, 1024], no_classes=1, output_size=(512, 512))
    optimizer = optim.SGD(model.parameters(), lr=1e-3, momentum=0.99)
    criterion = WeightedBCEWithLogitsLoss()

    # Loading weights
    start_epoch = 0
    if os.path.exists(weights_path):
        print("Loading weights...")
        checkpoint = torch.load(weights_path)
        if 'model_state_dict' in checkpoint:
            model.load_state_dict(checkpoint['model_state_dict'])
            optimizer.load_state_dict(checkpoint.get('optimizer_state_dict', optimizer.state_dict()))
            start_epoch = checkpoint.get('epoch', 0)
            print(f"Resuming training from epoch {start_epoch}...")
        else:
            model.load_state_dict(checkpoint)
            print("Loaded model weights only.")
            start_epoch = 0
    else:
        print("No weights found. Starting from scratch.")
        start_epoch = 0

    # Training
    train_unet(model, train_loader, criterion, optimizer, device, epochs=500, save_path=weights_path, start_epoch=start_epoch)



```
